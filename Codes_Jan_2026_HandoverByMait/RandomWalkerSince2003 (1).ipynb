```python
# Create an API object with your client credentials
da = deviantart.Api("42096", "97080792c6d30a4178965e41f1ca15de")

# Specify the username of the user whose profile you want to download
username = "DigitalAIHeaven"  # Replace with the desired username


    # Access and print specific profile data
# Access and print information about deviations in the gallery
#for deviation in gallery['results']:
 #   print(f"Deviation Title: {deviation.title}")
 #   print(f"Deviation ID: {deviation.deviationid}")
 #   print(f"Deviation URL: {deviation.url}")

 # Create a directory to store the downloaded images
output_dir = f"downloaded_images4_{username}"
if not os.path.exists(output_dir):
    os.makedirs(output_dir)

def download_image_metadata(image_url, output_dir, filename, metadata):
    """Downloads an image from a URL and saves it to a specified directory."""
    if not os.path.exists(output_dir):
        os.makedirs(output_dir)
    # Sanitize the filename by removing characters that are not allowed in filenames
    filename = "".join(x for x in filename if x.isalnum() or x in "._- ")
    try:
        response = requests.get(image_url, stream=True)
        response.raise_for_status()
        os.makedirs(output_dir, exist_ok=True)
        filename = os.path.join(output_dir, "metadata" + username + ".json")
    
        with open(filename, "w") as f:
            json.dump(metadata, f, indent=4)
    
        print(f"Saved metadata for deviation {username} to {filename}")
        #Download image files
        filepath = os.path.join(output_dir, filename)
        with open(filepath, 'wb') as image_file:
            for chunk in response.iter_content(chunk_size=8192):
                image_file.write(chunk)

        print(f"Downloaded: image {filepath}")

    except requests.exceptions.RequestException as e:
        print(f"Error downloading {image_url}: {e}")

# Get user's gallery folder
gallery = da.get_gallery_folder(username)
# Iterate through all deviations in the gallery
has_more = True
while has_more:
    for deviation in gallery['results']:
        metadata = {
        "title": deviation.title,
        "deviationid": deviation.deviationid,
        "url": deviation.url,
        "category": deviation.content
        # Add other metadata fields as needed
        }
        if hasattr(deviation, 'content') and deviation.content:
            if 'src' in deviation.content:
                image_url = deviation.content['src']
                filename = f"{deviation.title}-{deviation.deviationid}.jpg"
                download_image_metadata(metadata, image_url, output_dir, filename)

        print("-" * 20)

    # Check if there are more deviations to fetch
    has_more = gallery['has_more']
    if has_more:
        gallery = da.get_gallery_folder(username, offset=gallery['next_offset'])

print("Gallery download complete!")
```

```python
def get_response_rate(response):    
    if response.status_code == 200:
        return json.loads(response.content.decode('utf-8'))
    elif response.status_code == 404:
        return 'user_done'
    elif response.status_code == 429:
        return 'too_many_requests'
    elif response.status_code == 500:
        return 'server error'
    elif response.status_code == 401:
        return 'get new token'
```

```python
#Function to get friends
def get_friends(username, page):
    token=get_token()
    api_url=f"https://www.deviantart.com/api/v1/oauth2/user/friends/{username}?access_token={token}"
    response = requests.get(api_url, params={'offset': page, 'limit': 50})
    return response
```

```python
#Function to get watchers
def get_watchers(username, page):
    token=get_token()
    api_url=f"https://www.deviantart.com/api/v1/oauth2/user/watchers/{username}?access_token={token}"
    response = requests.get(api_url, params={'offset': page, 'limit': 50})
    return response
```
